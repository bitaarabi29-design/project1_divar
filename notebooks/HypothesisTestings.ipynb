```python
#3.داشتن سند تجاری (یا هر نوع سند ملکی) در املاک به این معنی است که سند مالکیت معتبر، رسمی و قانونی برای ملک تجاری دارید. این سند نشان می‌دهد که شما صاحب قانونی ملک تجاری هستید و می‌توانید از حقوق مالکیت آن استفاده کنید. بررسی کنید که آیا داشتن سند تجاری(has_business_deed) بر میانگین قیمت فروش ملک تجاری ارتباط معناداری دارد؟
#جدا سازی دو گروه
group_with_deed = df[df["has_business_deed"] == 1]["price_value"].dropna()
group_without_deed = df[df["has_business_deed"] == 0]["price_value"].dropna()
#میو:
print("has deed, mean:", group_with_deed.mean())
print("does not have deed, mean:", group_without_deed.mean())
#میانگین قیمت خانه های سند دار بیشتره
sample_with_deed = group_with_deed.sample(n=5000, random_state=44)
sample_without_deed = group_without_deed.sample(n=5000, random_state=44)

stat_deed, p_deed = stats.shapiro(sample_with_deed)
stat_no_deed, p_no_deed = stats.shapiro(sample_without_deed)
print(stat_deed, p_deed)
print(stat_no_deed, p_no_deed)

#پی ولیو هر دو کوچک تر از الفا که 5 صدم هست هستن که یعنی نرمال نیستن پس بریم سراغ من ویتنی
#بررسی برابری واریانس:

stat_levene, p_levene = stats.levene(sample_with_deed, sample_without_deed)
print(p_levene)
#کوچک تر از 5 صدم *الفا* بود پس واریانس ها برابر نیستن
u_stat, p_value = stats.mannwhitneyu(sample_with_deed, sample_without_deed, alternative='two-sided')
print(u_stat, p_value)
#پی ولیو کوچک تر از الفاست:
#فرض 0 رد می شود
#داشتن سند تجاری با میانگین قیمت فروش ملک تجاری ارتباط آماری معناداری داره.
```
