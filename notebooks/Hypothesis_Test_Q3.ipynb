## ۳. تأثیر داشتن سند تجاری بر قیمت فروش

داشتن سند تجاری (یا هر نوع سند ملکی) در املاک به این معنی است که **سند مالکیت معتبر، رسمی و قانونی** برای ملک تجاری دارید. این سند نشان می‌دهد که شما صاحب قانونی ملک تجاری هستید و می‌توانید از حقوق مالکیت آن استفاده کنید.  
بررسی کنید که آیا داشتن سند تجاری (`has_business_deed`) بر **میانگین قیمت فروش** ملک تجاری ارتباط معناداری دارد؟

### جداسازی دو گروه

داده‌ها به دو گروه **دارای سند** و **بدون سند** تقسیم شده‌اند.

### میانگین (μ) هر گروه

**میانگین قیمت** خانه‌های سنددار بیشتر است.

### بررسی نرمال بودن

پی‌ولیو هر دو گروه از **آلفا** که `0.05` است کوچک‌تر است، یعنی داده‌ها **نرمال نیستند**؛ پس سراغ آزمون **من‌ویتنی** می‌رویم.

### بررسی برابری واریانس

پی‌ولیو آزمون لوین از `0.05` (**آلفا**) کوچک‌تر بود، پس **واریانس‌ها برابر نیستند**.

### نتیجه آزمون من‌ویتنی

پی‌ولیو از آلفا کوچک‌تر است، بنابراین **فرض صفر رد می‌شود**.  
داشتن سند تجاری با میانگین قیمت فروش ملک تجاری **ارتباط آماری معناداری** دارد.

</div>

```python
group_with_deed = df[df["has_business_deed"] == 1]["price_value"].dropna()
group_without_deed = df[df["has_business_deed"] == 0]["price_value"].dropna()

print("has deed, mean:", group_with_deed.mean())
print("does not have deed, mean:", group_without_deed.mean())

sample_with_deed = group_with_deed.sample(n=5000, random_state=44)
sample_without_deed = group_without_deed.sample(n=5000, random_state=44)

stat_deed, p_deed = stats.shapiro(sample_with_deed)
stat_no_deed, p_no_deed = stats.shapiro(sample_without_deed)
print(stat_deed, p_deed)
print(stat_no_deed, p_no_deed)

u_stat, p_value = stats.mannwhitneyu(sample_with_deed, sample_without_deed, alternative='two-sided')
print(u_stat, p_value)
```
